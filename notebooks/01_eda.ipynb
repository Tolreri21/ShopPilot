## Short conclusions

**General**
- Neither `examples` nor `products` has duplicates (the shape does not change after `drop_duplicates`).
- The two tables are linked: `examples` tells us which product was shown for which query, and `products` holds the product details. We join them by `product_id`.

**`examples`**
- Two columns change the data a lot: `product_locale` and `small_version`. We will use them in preprocessing (keep only `us` and the small subset).
- The `split` column already says what is train and what is test, so we do not need to invent our own split.
- The share of `E` (exact) labels is different: about 0.68 in the full dataset, but only about 0.43 in the small one. So the small version is not a random copy of the full one, and its label balance is different.
- A query has about 18 rows on average (18.7), and each row is a different product.

**`products`**
- There are many missing values.
  - `product_description` is missing in about half of the rows, so it is not reliable.
  - `product_bullet_point` and `product_brand` are missing much less (up to about 20%), so we will search by them.
  - `product_color` is missing in about 400k rows, so it is not very useful either.
- Columns we will probably keep:
  - `product_id` to join with `examples`
  - `product_title`
  - `product_bullet_point`
  - `product_brand`
  - (title, bullet point and brand give the context for each product)
